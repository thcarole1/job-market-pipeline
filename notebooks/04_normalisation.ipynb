```python
import json
import pandas as pd

# Afficher toutes les colonnes (sans limite)
pd.set_option("display.max_columns", None)
```

## Chargement des données

```python
def charger_offres(path):
    '''
    Charge un fichier de données en format json à partir de sa localisation (chemin).
    Renvoie une liste.
    '''
    with open(path, "r", encoding="utf-8") as f:
        offres = json.load(f)
        return offres

# fichiers de données 
path_ft = "../data/processed/francetravail/offres_20260401_092159.json"
path_wttj = "../data/processed/welcometothejungle/offres_20260401_092124.json"

# charger offres France travail et Welcome to the jungle
offres_ft = charger_offres(path_ft)
offres_wttj = charger_offres(path_wttj)

print(f"{len(offres_ft)} offres FT chargées")
print(f"{len(offres_wttj)} offres WTTJ chargées")

# Chargement des données dans une dataframe
offres_ft_df = pd.DataFrame(offres_ft)
offres_wttj_df = pd.DataFrame(offres_wttj)
```

## Exploration des données

```python
def rechercher_duplicates(df):
    # Check number of rows before removing duplicates
    print(f"Number of rows : {len(df)}\n")

    # # Compute the number of duplicated rows
    # print(f"Number of duplicated rows : {df.duplicated().sum()}\n")

def get_missing_data(data):       
    # Counting the percentage of NaN for each column
    missing_data = data.isnull().sum().sort_values(ascending=False)/len(data)*100 #NaN percentage for each column

    # missing_data
    missing_data = missing_data[missing_data > 0]

    #Missing data column names
    missing_cols = missing_data.index.values.tolist()

    #Show missing data in percent
    if len(missing_data)>0:
        print("missing data (isnull) detected are as follows:")
        display(missing_data)
    else:
        display("no missing data detected\n")

def explore(df):
    #Numerical Features
    numerical_features = df.select_dtypes(include = 'number').columns.values.tolist()
    
    print(f"Shape is : {df.shape}\n")

    display(df.sample(5))

    display(df.info())
    print("\n")
    
    if numerical_features:
        print("Description is : \n")
        display(df.describe())
    else : 
        print("No numerical features to describe.\n")

    get_missing_data(df)
```

### France Travail

```python
# Champs obligatoires
champs_obligatoires_ft = ["id", "source","titre", "entreprise","description",
                       "competences","localisation_ville","type_contrat",
                       "type_contrat_libelle","date_publication","url"]

# Champs recommandés (il manque missions et télétravail chea france travail)
champs_recommandes_ft = ["salaire_min", "salaire_max","experience_min",
                      "experience_exige","teletravail","secteur",
                      "localisation_dept","latitude","longitude","nb_employes"]


print("\n*********************** CHAMPS OBLIGATOIRES FT  *************************")
df_obligatoire_ft = offres_ft_df[champs_obligatoires_ft]
explore(df_obligatoire_ft)
print("\n***********************************************************************\n")

print(" \n*********************** CHAMPS RECOMMANDES FT *************************")
df_reco_ft = offres_ft_df[champs_recommandes_ft]
explore(df_reco_ft)
print("\n***********************************************************************\n")

# df_reco_ft.duplicated()
# df_obligatoire_ft.duplicated()
```

```python
# Champs obligatoires (il manque le champ compétences à WTTJ)
champs_obligatoires_wttj = ["id","source","titre","entreprise","description",
                            "localisation_ville","type_contrat",
                            "date_publication","url"]

# Champs recommandés (il manque missions et télétravail chea france travail)
champs_recommandes_wttj = ["salaire_min", "salaire_max","experience_min",
                           "teletravail","secteur","localisation_region",
                           "latitude","longitude","nb_employes","missions"]



print(" *********************** CHAMPS OBLIGATOIRES WTTJ  *************************")
df_obligatoire_wttj = offres_wttj_df[champs_obligatoires_wttj]
explore(df_obligatoire_wttj)
print("\n***********************************************************************\n")

print(" *********************** CHAMPS RECOMMANDES WTTJ *************************")
df_reco_wttj = offres_wttj_df[champs_recommandes_wttj]
explore(df_reco_wttj)
print("\n***********************************************************************\n")
```
